```python
import pandas as pd
import numpy as numpy
import os
import glob
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 50)
```

```python
DATA_DIR = os.path.join('..', "datasets")
csv_files = glob.glob(os.path.join(DATA_DIR, "*.csv"))
print('Total csv files found : ', len(csv_files))
```

```python
file_summaries = []
for file_path in csv_files:
    file_name = os.path.basename(file_path)
    size_mb = os.path.getsize(file_path) / (1024 * 1024)
    
    # Read only 5 rows for super fast inspection
    sample_df = pd.read_csv(file_path, nrows=5)
    
    file_summaries.append({
        "File Name": file_name,
        "Size (MB)": round(size_mb, 2),
        "Columns Count": len(sample_df.columns)
    })
summary_table = pd.DataFrame(file_summaries)
summary_table
```

```python
label_counts_per_file = {}
for file_path in csv_files:
    file_name = os.path.basename(file_path)
    sample = pd.read_csv(file_path, nrows=1)
    sample.columns = sample.columns.str.strip()
    label_counts_per_file[file_name] = sample["Label"].value_counts()
```

```python
label_counts_per_file = {}

for file_path in csv_files:
    file_name = os.path.basename(file_path)
    
    # Read the label column by stripping spaces from column headers
    df_label = pd.read_csv(
        file_path, 
        usecols=lambda col: col.strip() == "Label",
        encoding="cp1252"
    )
    df_label.columns = df_label.columns.str.strip()
    
    label_counts_per_file[file_name] = df_label["Label"].value_counts()

# Print full attack distribution
for file_name, counts in label_counts_per_file.items():
    print(f"=== {file_name} ===")
    print(counts)
    print()
```
